## Step 7: Missing Values

```python
#converting any placeholders for missing/null data into detectable null data
df["smoking_status"] = df["smoking_status"].replace("Unknown", np.nan)

#identifying any missing and null values in the dataset
missing_count = df.isnull().sum()


percent_missing = missing_count * 100/len(df)

print("-----------------------------------------------")
print("Columns with missing or null values")
print("-----------------------------------------------")
print(missing_count)
print("-----------------------------------------------")
print("Percentage of each column that is missing data")
print("-----------------------------------------------")
print(percent_missing)
print("-----------------------------------------------")
```

### Findings: 
Which columns have missing values (according to .isnull())? The "bmi" and "smoking_status" columns have missing values.

What percentage of each column is missing? 3.93% of data in the "bmi" column is missing, and 30.2% in the "smoking_status" column is missing. 

⚠️ IMPORTANT: Are there any values that look like data but actually represent missing information? (Hint: Look carefully at the actual values in each column) Yes. Originally, missing values for the "smoking_status" column were reported as "Unknown." To be detected by pandas, these values were converted to NaN values.
