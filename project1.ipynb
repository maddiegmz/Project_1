## Step 7: Missing Values

```python
#identifying any missing and null values in the dataset
missing_count = df.isnull().sum()
percent_missing = missing_count * 100/len(df)

print("-----------------------------------------------")
print("Columns with missing or null values")
print("-----------------------------------------------")
print(missing_count)
print("-----------------------------------------------")
print("Percentage of each column that is missing data")
print("-----------------------------------------------")
print(percent_missing)
print("-----------------------------------------------")
```

## Step 8: Duplicates

```python
#identifying any duplicates in the dataset
num_duplicates = df.duplicated().sum()
id_duplicates = df.duplicated(subset = "id").sum()

print("-----------------------------------------------")
print(f"There are {num_duplicates} duplicates in this dataset.")
print(f"There are {id_duplicates} duplicates in the 'id' column.")
print("-----------------------------------------------")
```
